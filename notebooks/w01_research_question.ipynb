## 2) The question: decision, action, and cost of a wrong call

### Research question

Which pages appear to have a meaningful opportunity to improve organic
click-through performance, based on their available search-performance signals?

### Unit of analysis

The unit of analysis is a **page**.

Each row represents a page-level observation and its associated search-performance
features.

### Decision

The decision is whether a page should be prioritized for further investigation
or possible content/search-snippet improvement.

### Output

The output will eventually be an opportunity score or ranked list of pages that
appear more likely to have improvement potential.

### Action

An SEO/content team could use the output to decide which pages to review first.
For example, they could inspect the page's title, description, content alignment,
search intent, and query coverage before deciding whether to make an update.

### Cost of a wrong call

A false positive could cause the team to spend time improving a page that has
little opportunity for improvement.

A false negative could cause a genuinely useful page to be overlooked, potentially
leaving an improvement opportunity unexplored.

Therefore, the model should support prioritization rather than automatically
deciding that a page must be changed.

The decision this project improves is: which content pages should be reviewed first for potential refresh opportunities.

The person acting on the output is a content strategist or SEO specialist. They can use the ranked list to decide which pages need investigation, such as updating content, improving engagement, or monitoring performance.

A wrong recommendation can waste editing time on pages that do not need improvement, while missing pages that have real opportunities. Therefore, the system should prioritize accurate ranking and provide explanations for why a page was selected.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

df = pd.read_csv("content_refresh_anonymized.csv")

print("Number of content items:", len(df))
print("Number of clients:", df["client_id"].nunique())

print("\nTrend distribution:")
print(df["trend_direction"].value_counts())
```

```python
# Show columns so we can identify the relevant CTR/opportunity fields

print(df.columns.tolist())
```

The dataset contains 30,000 anonymized content items across 32 clients.

The starter dataset includes observable search and engagement signals such as impressions, CTR, average position, sessions, and trend measurements.

A first observation is that content performance is not uniform: pages have different trend directions, meaning some pages are improving while others are declining. This supports the need for a ranking system that helps prioritize review efforts.

```python
print("Number of rows:", len(df))

if "ctr_gap" in df.columns:
    print("CTR gap non-null values:", df["ctr_gap"].notna().sum())
    print("Average CTR gap:", round(df["ctr_gap"].mean(), 4))
    print("Median CTR gap:", round(df["ctr_gap"].median(), 4))
```

```python
declining_rate = (df["trend_direction"] == "down").mean()

print("Percentage of declining pages:", round(declining_rate * 100, 2), "%")
```

Around 54.21% of pages show a downward trend in the dataset, indicating that identifying which pages require attention is a meaningful prioritization problem.
