## 3. Success metric

The main success metric will be Precision@K, where K represents the number of top-ranked pages that the content or SEO team can realistically review.

Precision@K measures how many of the top K pages identified by the model are genuinely useful opportunities.

This metric fits the task because the main goal is to create a useful priority list rather than make a binary prediction for every page. A high Precision@K would mean that the pages placed near the top of the review queue are more likely to be valuable opportunities.

The current starter data does not contain verified future outcomes, so Precision@K cannot be calculated yet. It would be evaluated later when labeled outcomes or human-reviewed opportunity labels are available.## 3. Success metric
